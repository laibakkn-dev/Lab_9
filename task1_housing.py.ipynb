```python
import pandas as pd
from sklearn.linear_model import LinearRegression

homes = pd.read_csv("real_estate.csv")
homes["distance_100m"] = homes["X3 distance to the nearest MRT station"] / 100
X1 = homes[["distance_100m"]]
y1 = homes["Y house price of unit area"]

# (a) fit
m1 = LinearRegression().fit(X1, y1)
slope, intercept = m1.coef_[0], m1.intercept_
print("Rows:", len(homes))
print(f"Slope:     {slope:.4f} price units per 100 m")
print(f"Intercept: {intercept:.4f} price units")

# (b) predict at 500 m  ->  distance_100m = 5
pred_500 = m1.predict(pd.DataFrame({"distance_100m": [5]}))[0]
print(f"Predicted price at 500 m: {pred_500:.2f} price units")
```
